```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


data = {
    'Qubits': [4, 6, 8, 10, 12],
    'Accuracy_mean': [0.9812, 0.9844, 0.9969, 1.0000, 0.9969],
    'Accuracy_std': [0.0063, 0.0000, 0.0063, 0.0000, 0.0063],
    'AUC_mean': [0.9985, 0.9999, 1.0000, 1.0000, 1.0000],
    'AUC_std': [0.0010, 0.0001, 0.0000, 0.0000, 0.0000],
    'F1_mean': [0.9809, 0.9843, 0.9969, 1.0000, 0.9969],
    'F1_std': [0.0068, 0.0000, 0.0063, 0.0000, 0.0063]
}

df = pd.DataFrame(data)


plt.figure(figsize=(11, 7))

# --- Define custom colors and markers ---
metrics = {
    'Accuracy': {'color': '#1f77b4', 'marker': 'o'},   # Blue
    'Macro AUC': {'color': '#2ca02c', 'marker': 's'},  # Green
    'Macro F1': {'color': '#d62728', 'marker': '^'}    # Red
}

# --- Plot Accuracy ---
plt.plot(df['Qubits'], df['Accuracy_mean'], '-o',
         color=metrics['Accuracy']['color'], markersize=8, linewidth=2,
         label='Accuracy (Mean)')
plt.fill_between(df['Qubits'],
                 df['Accuracy_mean'] - df['Accuracy_std'],
                 df['Accuracy_mean'] + df['Accuracy_std'],
                 color=metrics['Accuracy']['color'], alpha=0.2)

# --- Plot Macro AUC ---
plt.plot(df['Qubits'], df['AUC_mean'], '-s',
         color=metrics['Macro AUC']['color'], markersize=8, linewidth=2,
         label='Macro AUC (Mean)')
plt.fill_between(df['Qubits'],
                 df['AUC_mean'] - df['AUC_std'],
                 df['AUC_mean'] + df['AUC_std'],
                 color=metrics['Macro AUC']['color'], alpha=0.2)

# --- Plot Macro F1 ---
plt.plot(df['Qubits'], df['F1_mean'], '-^',
         color=metrics['Macro F1']['color'], markersize=8, linewidth=2,
         label='Macro F1-score (Mean)')
plt.fill_between(df['Qubits'],
                 df['F1_mean'] - df['F1_std'],
                 df['F1_mean'] + df['F1_std'],
                 color=metrics['Macro F1']['color'], alpha=0.2)


plt.xlabel('Number of Qubits', fontsize=14)
plt.ylabel('Performance Metrics (Mean ± Std)', fontsize=14)
plt.ylim(0.95, 1.009)
plt.xticks(df['Qubits'], fontsize=12)
plt.yticks(fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)


plt.legend(
    title='Performance Metrics',
    title_fontsize=13,
    fontsize=12,
    loc='lower right',
    frameon=True,
    shadow=True,
    fancybox=True,
    borderpad=1,
    labelspacing=0.8
)

plt.tight_layout()
plt.savefig('acc1.pdf')
plt.show()
```
